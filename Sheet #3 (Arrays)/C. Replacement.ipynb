```python
n = int(input())
a = list(map(int,input().split()))
for i in a :
    if i >0:
        print(1,end=" ")
    elif i<0:
        print(2,end=" ")
    else :
        print(0,end=" ")
```

## Another solution

```python
n = int(input())
a = list(map(int,input().split()))
res=[]
for i in a:
    if i >0 :
        res.append("1")
    elif i<0 :
        res.append("2")
    else :
        res.append("0")
print (" ".join(res))
```
