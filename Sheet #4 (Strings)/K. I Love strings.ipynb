```python
ran = int(input())

for i in range(ran):
    s, t = input().split()

    res = []
    min_len = min(len(s), len(t))

    for j in range(min_len):
        res.append(s[j])
        res.append(t[j])

    res.append(s[min_len:])
    res.append(t[min_len:])

    print("".join(res))
```
